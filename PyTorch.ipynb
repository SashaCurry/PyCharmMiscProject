```python
tensor = torch.tensor([1, 2, 3 ,4])

# VIEW создаёт представление данных, не делая копию. Работает только если тензор лежит последовательно в памяти
tensor_1 = tensor.view([4 ,1])

# Сначала пытается сделать представление. Если не получается (данные лежат не последовательно), создаёт копию
tensor_2 = tensor.reshape([2, 2])

print(tensor)
print(tensor_1)
print(tensor_2)
```

```python
# Добавление осей

tensor = torch.unsqueeze(tensor, 0)

tensor.shape
```

```python
# Арифметические операции

tensor = torch.tensor([1, 2, 3, 4, 5])

tensor * 5

tensor + torch.tensor([1, 2, 3, 4, 5])

# tensor + [1, 2, 3, 4, 5]
```

```python
# Математические функции

tensor = torch.tensor([1, 2, 3, 4, 5, 6], dtype=torch.float32)

# tensor.sum()
# tensor.mean()

tensor = tensor.view([2, 3])
tensor_mean = tensor.mean(dim=1, keepdim=True)

print(tensor)
print(tensor_mean)
print(tensor_mean.shape)
```

```python
tensor = torch.tensor([1., 2., 3.], requires_grad=True)
```

```python
# tensor = tensor.cpu()
# tensor = tensor.to('cpu')

tensor = tensor.cuda()
# tensor = tensor.cuda('cuda')

tensor
```

```python
# PATTERN

device = 'cuda' if torch.cuda.is_available() else 'cpu'

tensor = tensor.to(device)
tensor
```

```python
# Check CPU or GPU

tensor.device
```

```python
tensor_cpu = torch.tensor([1., 2., 3.]).to('cpu')

tensor_cpu
```

```python
# Тезоры нельзя складывать, когда они нга разных девайсах

# tensor + tensor_cpu

new_tensor = tensor + tensor_cpu.to(device)

new_tensor
```

```python
# Тензоры копируются вместе с градиентом

new_tensor = new_tensor.cpu()

new_tensor = new_tensor.cpu().detach()

new_tensor
```
